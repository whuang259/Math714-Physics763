# Problem 1

We first derive the analytical solution for a 2D Poiseuille flow in a rectangular channel. Consider an incompressible, Newtonian fluid flowing steadily between two parallel plates separated by a distance $H$. The plates are stationary, and no-slip boundary conditions hold at both walls. The flow is driven solely by a constant pressure gradient in the $x$-direction. Gravity and all other body forces are neglected.

We use the following assumptions:
* steady flow: $\partial u/\partial t = 0$
* fully developed flow: $\partial u/\partial x = 0$
* unidirectional flow: $\mathbf{u}(x, y) = (u(y), 0)$

1. Using the assumptions above, simplify the Navier–Stokes equations and show that the only nontrivial component is the $x$-momentum equation:
$$ \mu \frac{d^2 u}{dy^2} = \frac{dp}{dx} $$ 
where $dp/dx$ is constant.

2. Find the general form of the velocity profile $u(y)$ by integrating the above equation twice. You can use the no-slip boundary conditions $u(0) = 0$ and $u(H) = 0$ to determine the integration constants.

3. Derive the final expression for the pressure-driven Poiseuille flow:
$$ u(y) = \frac{1}{2\mu} \left(-\frac{dp}{dx} \right) y(H- y) $$

4. Determine the location and value of the maximum velocity $u_{max}$.

### Part 1

The Navier-Stokes equations are
$$ \nabla \cdot \mathbf{u} = 0 \qquad \frac{\partial \mathbf{u}}{\partial t} = -(\mathbf{u} \cdot \nabla)\mathbf{u} - \frac{\nabla p}{\rho} + v \nabla^2 \mathbf{u} + \mathbf{f} $$

We apply the stated assumptions to simplify the second equation as
$$ 0 = - \frac{\nabla p}{\rho} + v u_{yy} + \mathbf{f}$$

We also have the there are no body forces and flow is driven by a gradient in the $x$-direction, so the equation becomes
\begin{gather*}
    0 = -\frac{p_x}{\rho} + v u_{yy} \\
    v\rho u_{yy} = \mu u_{yy} = \frac{dp}{dx} 
\end{gather*}

The mass conservation equation becomes
$$ u_x + u_y = u_y = 0 $$

### Part 2/3

Since $dp/dx$ is constant, we integrate twice to get
$$ \mu u(y) = \frac{1}{2} p_x y^2 + C_0 y + C_1 $$

The boundary conditions gives $C_1 = 0$ and $C_0 = - \frac{1}{2} p_x H$, thus the general form is
$$ u(y) = \frac{1}{2\mu} \frac{dp}{dx} \left( y^2 - H y \right) $$

We can rewrite this in the desired form
$$ u(y) = \frac{1}{2\mu} \left( -\frac{dp}{dx} \right) y(H - y) $$

### Part 4

We can take the first derivative to see that the maximum occurs at $y = H/2$ and
$$ u(H/2) = -\frac{H^2}{4\mu} \left( \frac{dp}{dx} \right) $$
